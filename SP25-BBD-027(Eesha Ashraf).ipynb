```python
#PART D
# Analyze a dataset of customer orders to find the most popular product category.
orders = [
    {'category': 'Electronics'},
    {'category': 'Clothing'}, 
    {'category': 'Electronics'},
    {'category': 'Food'}
]
categories = [item['category'] for item in orders]
popular = max(set(categories), key=categories.count)
print(popular)
```

```python
# PART F
#Inventory Management:
#Create a dictionary to store product information: name, quantity, price.
#Implement functions to:
#Add new products
#Update product quantities
#Calculate total inventory value
#Generate a report of low-stock items
inventory = {}
def add(name, qty, price):
    inventory[name] = {'qty': qty, 'price': price}
def update(name, change):
    inventory[name]['qty'] += change
def total():
    return sum(item['qty'] * item['price'] for item in inventory.values())
def low_stock():
    return {name: data for name, data in inventory.items() if data['qty'] < 5}
add('Apple', 10, 1.5)
add('Banana', 3, 0.75)
update('Apple', -3)
print(inventory)
print(total())
print(low_stock())
```

```python
#PART G
# Student Gradebook:
#Create a dictionary to store student information: name, grades for different subjects.
#Calculate the average grade for each student.
#Find the highest and lowest grades in a specific subject.
#Sort students by their average grades.
grades = {
    "Alice": {"Math": 85, "Science": 90},
    "Bob": {"Math": 78, "Science": 82},
    "Charlie": {"Math": 92, "Science": 88}
}
# Find averages
averages = {}
for name in grades:
    scores = grades[name].values()
    averages[name] = sum(scores) / len(scores)
# Check math scores
math_scores = [grades[name]["Math"] for name in grades]
best_math = max(math_scores)
worst_math = min(math_scores)
# Sort students
sorted_students = sorted(averages.items(), key=lambda x: x[1], reverse=True)
print("Student averages:", averages)
print("Best math score:", best_math)
print("Worst math score:", worst_math)
print("Class ranking:", sorted_students)
```
